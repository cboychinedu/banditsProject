```python
# Importing the necessary modules 
import cv2 
import torch 
import supervision as sv 
from ultralytics import YOLO 
from IPython.display import Video
```

```python
# Setting the path to the trained model 
modelPath = "model/best.pt" 

# Loading the model into memory 
model = YOLO(modelPath) 
```

```python
# Displaying the video 
Video("video.mp4", width=800, height=600)
```

```python
# Setting the path to the video 
videoPath = "video.mp4" 

# Reading the video into memory
cap = cv2.VideoCapture(videoPath) 

# Get metadata for the VideoWriter
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Define the video codec and create VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*'avc1')
out = cv2.VideoWriter('analyzedVideo.mp4', fourcc, fps, (frame_width, frame_height))
```

```python
# Perfoming the machine learning analysis 
while cap.isOpened(): 
    # Read each frames
    ret, frame = cap.read()

    # if not ret, break 
    if not ret: 
        break 

    # Getting the results 
    results = model(frame)[0]

    # Getting the detections 
    detections = sv.Detections.from_ultralytics(results) 

    # Drawing the box 
    boxAnnotator = sv.BoxAnnotator() 
    labelAnnotator = sv.LabelAnnotator() 

    # Getting the labels 
    labels = [
        f"{className} {confidence:.2f}"
        for className, confidence 
        in zip(detections["class_name"], detections.confidence)
    ]

    # Getting the anotated image 
    annotatedImage = boxAnnotator.annotate(
        scene=frame, detections=detections)

    # Adding the label 
    annotatedImage = labelAnnotator.annotate(
        scene=annotatedImage, detections=detections, labels=labels)

    # Writing the frame to disk 
    out.write(annotatedImage)

# Releaase resources when finished 
cap.release() 
out.release() 
cv2.destroyAllWindows() 
```

```python
# completed
print("Video Analysis Complete!") 
```
